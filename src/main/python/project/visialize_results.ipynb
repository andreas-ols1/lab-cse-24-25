# Visualize results

```python
import pandas as pd
```

```python

with open("output/part-00000", 'r') as f:
    output = f.readlines()

formatted_data = [
    [item if i < 2 else f"{float(item):.1f}"
    for i, item in enumerate(line.strip().strip("()").split(","))]
    for line in output
]

df = pd.DataFrame(formatted_data, columns=["Carrier", "Airport", "Delay (minutes)"])

# Find the most common airport(s)
airport_counts = df["Airport"].value_counts()
max_count = airport_counts.max()
most_common_airports = airport_counts[airport_counts == max_count]
print("Most occurring airport(s):")
for airport, count in most_common_airports.items():
    print(f"{airport} with {count} occurrences")
print("")

# Find the max and min delay
df["Delay (minutes)"] = df["Delay (minutes)"].astype(float)

max_delay_idx = df["Delay (minutes)"].idxmax()
max_delay_pair = df.loc[max_delay_idx, ["Carrier", "Airport", "Delay (minutes)"]]

min_delay_idx = df["Delay (minutes)"].idxmin()
min_delay_pair = df.loc[min_delay_idx, ["Carrier", "Airport", "Delay (minutes)"]]

print("Carrier-Airport pair with Most Delay:")
print(max_delay_pair.to_string(index=False))

print("\nCarrier-Airport pair with Least Delay:")
print(min_delay_pair.to_string(index=False))

# Print the dataframe
df
```
